# **Testing Feasible Shapley Values on a Known Causal Structure**
**A Monte Carlo Study on the Diamond DAG with Linear and Nonlinear Data-Generating Processes**

```python
# Import required libararies
import numpy as np
import pandas as pd
```

```python
# Data generatetion functions
def gen_diamond(n, a, b, nonlin=False, rng=None):
    rng = rng or np.random.default_rng()
    X1 = rng.normal(size=n)
    X2 = a*X1 + rng.normal(size=n)
    X3 = a*X1 + rng.normal(size=n)
    X4 = a*X1 + rng.normal(size=n)
    noise = rng.normal(size=n)
    Y = (b*X2**2 if nonlin else b*X2) + b*X3 + b*X4 + noise
    return pd.DataFrame({"X1":X1,"X2":X2,"X3":X3,"X4":X4,"Y":Y})
```

For the first phase:
- To compares linear vs nonlinear behaviour, the same-seed design is cleaner to explain (“same inputs, only the functional form changes”).
- For two independent test cases, keep different seeds.

For the 100-replication Monte Carlo:
- Give each replication its own reproducible seed, for example rng = np.random.default_rng(42 + r) for r = 0…99, or more rigorously np.random.SeedSequence(42).spawn(100). Otherwise every replication would produce the same dataset, and the 100 runs would tell you nothing about noise.

```python
# Generate datasets
dfA_linear = gen_diamond(2000, 0.7, 3, nonlin=False, rng=np.random.default_rng(42))
dfB_nonlinear = gen_diamond(2000, 0.7, 3, nonlin=True,  rng=np.random.default_rng(42))

# Save the dataset in CSV
dfA_linear.to_csv("dfA_linear.csv", index=False)
dfB_nonlinear.to_csv("dfB_nonlinear.csv", index=False)

# print the datasets
print
```
